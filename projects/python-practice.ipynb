QUESTION *1* - Temperature Analysis

```python

temperatures = [18, 22, 19, 25, 17, 21, 23, 20, 26, 24,
                19, 22, 27, 21, 23, 18, 24, 25, 19, 20,
                22, 26, 23, 21, 19, 24, 22, 20, 18, 23]

hot_days = 0

for temp in temperatures:
    if temp > 24:
        hot_days += 1

print("Number of hot days (above 24°C):", hot_days)
print("Highest temperature:", max(temperatures), "°C")
print("Lowest temperature:", min(temperatures), "°C")
```

QUESTION 2- Student Grades Processing

```python

scores = [88, 76, 90, 85, 92, 67, 73, 81, 95, 78]

average = sum(scores) / len(scores)
print("Average score:", average)

passed = 0
failed = 0

for score in scores:
    if score >= 70:
        passed += 1
    else:
        failed += 1

print("Number of students who passed:", passed)
print("Number of students who failed:", failed)

for i in range(len(scores)):
    if scores[i] >= 70:
        print("Student", i + 1, ": Pass")
    else:
        print("Student", i + 1, ": Fail")
```

QUESTION 3- Inventory Management

```python

inventory = [
    {'item': 'Apple', 'quantity': 50, 'price': 0.5},
    {'item': 'Banana', 'quantity': 20, 'price': 0.3},
    {'item': 'Orange', 'quantity': 30, 'price': 0.7},
    {'item': 'Grapes', 'quantity': 15, 'price': 1.5},
    {'item': 'Mango', 'quantity': 10, 'price': 2.0},
]

# Part A: Find items that need restocking
restock = []

for item in inventory:
    if item['quantity'] <= 20:
        restock.append(item['item'])

print("Items to restock:", ", ".join(restock))

# Part B: Calculate total inventory value
total = 0

for item in inventory:
    total += item['quantity'] * item['price']

print("Total inventory value: $", format(total, ".2f"), sep="")

# Part C: Increase prices by 10%
print("Updated Inventory Prices:")

for item in inventory:
    if item['quantity'] <= 15:
        item['price'] = item['price'] * 1.10
        print(item['item'] + ": $" + format(item['price'], ".2f"))
```

QUESTION 5: Text Processing


```python

words = ['python', 'data', 'science', 'machine', 'learning', 'artificial', 'intelligence']

# Part A: Find the length of each word
lengths = []

for word in words:
    lengths.append(len(word))

print("Lengths of words:", lengths)

# Part B: Find the longest word
longest_word = max(words, key=len)

print("Longest word:", repr(longest_word))

# Part C: Find words starting with 'a'
a_words = []

for word in words:
    if word.startswith('a'):
        a_words.append(word)

print("Words starting with 'a':", a_words)
```

QUESTION 6: Matrix Addition


```python

matrix_a = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

matrix_b = [
    [9, 8, 7],
    [6, 5, 4],
    [3, 2, 1]
]

# Part A: Add the two matrices
result = []

for i in range(3):
    row = []
    for j in range(3):
        row.append(matrix_a[i][j] + matrix_b[i][j])
    result.append(row)

# Part B: Print the resulting matrix
print("Resulting Matrix:")

for row in result:
    print(row)
```

QUESTION 7: Fetching Random Jokes


```python

import requests

# Part A: Fetch a random joke
url = "https://official-joke-api.appspot.com/random_joke"

response = requests.get(url)
joke = response.json()

# Part B: Print the joke and punchline
setup = joke["setup"]
punchline = joke["punchline"]

print("Joke:", setup)
print("Punchline:", punchline)

# Part C: Save the joke to a text file
with open("jokes.txt", "w") as file:
    file.write("Joke: " + setup + "\n")
    file.write("Punchline: " + punchline)
```

QUESTION 8: Advanced - News Aggregator with Database Storage


```python

import requests
import sqlite3

# Part A: Ask the user for a news topic
topic = input("Enter news topic: ")

# Connect to the database
conn = sqlite3.connect("news.db")
cursor = conn.cursor()

# Create a table for news
cursor.execute("""
CREATE TABLE IF NOT EXISTS news (
    title TEXT UNIQUE,
    description TEXT,
    published_at TEXT
)
""")

# Part B: Get the top 5 news headlines
api_key = "YOUR_API_KEY"

url = "https://newsapi.org/v2/everything"

params = {
    "q": topic,
    "apiKey": api_key,
    "pageSize": 5,
    "sortBy": "publishedAt"
}

response = requests.get(url, params=params)
data = response.json()

if response.status_code == 200:
    articles = data["articles"]

    # Part C and E: Save news and avoid duplicates
    for article in articles:
        title = article["title"]
        description = article["description"]
        published_at = article["publishedAt"]

        cursor.execute("""
        INSERT OR IGNORE INTO news
        VALUES (?, ?, ?)
        """, (title, description, published_at))

    conn.commit()

    # Part D: Read and display news
    cursor.execute("""
    SELECT title, description, published_at
    FROM news
    ORDER BY published_at DESC
    LIMIT 5
    """)

    news = cursor.fetchall()

    print("\nTop Headlines:")

    for i, article in enumerate(news, 1):
        print("\n", i, ". Title: ", article[0], sep="")
        print("Description:", article[1])
        print("Published At:", article[2][:10])

    print("\nData has been saved to the database.")

else:
    print("Error:", data.get("message", "Unable to fetch news"))

conn.close()
```
